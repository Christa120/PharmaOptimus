## 1. Chargement et Audit des Données
Analyse des zéros, de l'intermittence et de la censure de consommation (`stockout_days`).


```python
import pandas as pd
import numpy as np

df = pd.read_csv('../../data/processed/historical_consumption_benin.csv')
print('Nombre d\'enregistrements :', len(df))
print('Taux de semaines à zéro :', round((df['qty_dispensed'] == 0).mean() * 100, 2), '%')
print('Semaines avec rupture constatée (stockout_days > 0) :', (df['stockout_days'] > 0).sum())
```

## 2. Détection de la Demande Censurée
En période de rupture (`stockout_days > 0`), la demande observée est inférieure à la demande réelle.


```python
censored = df[df['stockout_days'] > 0]
print('Impact moyen de la censure :', censored['stockout_days'].mean(), 'jours/semaine')
```
